```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler, LabelEncoder
```

```python
iris=load_iris(as_frame=True)
df=iris.frame
df.head()
```

```python
print("Shape of dataset: ", df.shape)
```

```python
print("Column Names")
print(df.columns.to_list())
```

```python
print("Dataset information")
print(df.info())
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
print("Total missing vales")
print(df.isnull().sum().sum())
```

```python
print(df.dtypes)
```

```python
print("Integer columns: ",df.select_dtypes(include=["int"]).columns.to_list())
print("Float columns: ",df.select_dtypes(include=["float"]).columns.to_list())
print("Character columns:", df.select_dtypes(include=["object"]).columns.tolist())
print("Logical columns:", df.select_dtypes(include=["bool"]).columns.tolist())
```

```python
df["species"]=df["target"].map(dict(enumerate(iris.target_names)))
df=df.drop(columns=["target"])
df["species"]=df["species"].astype("category")
print(df.dtypes)
display(df.head())
```

```python
print("Duplicate values: ",df.duplicated().sum())
```

```python
df.drop_duplicates()
```

```python
print("after removing duplicate values: ",df.shape)
```

```python
df.columns=(
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ","_")
        .str.replace(r"[^\w]","",regex=True)
)
print(df.columns.tolist())
print(df.head())
```

```python
encoder=LabelEncoder()
df["species_encoded"]=encoder.fit_transform(df["species"])
print("species encoding:")
print(dict(zip(encoder.classes_,encoder.transform(encoder.classes_))))
display(df[["species","species_encoded"]].head(10))
```

```python
features=[
    "sepal_length_cm",
    "sepal_width_cm",
    "petal_length_cm",
    "petal_width_cm"
]
scaler=MinMaxScaler()
df_normalized=df.copy()
df_normalized[features]=scaler.fit_transform(df_normalized[features])
display(df_normalized.head())
```

```python
print("Minimum values:")
print(df_normalized[features].min())
```

```python
print("Maximum values:")
print(df_normalized[features].max())
```

```python
print("Final normalized shape:", df_normalized.shape)
print()
```
