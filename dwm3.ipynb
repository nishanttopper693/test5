```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
transactions = [
    ["Python Programming", "Data Science", "Machine Learning"],
    ["Python Programming", "Data Science"],
    ["Python Programming", "Machine Learning"],
    ["Python Programming", "Data Science", "Machine Learning"],
    ["Data Science", "Machine Learning"],
    ["Python Programming", "Data Science"],
    ["Python Programming", "Data Science", "Machine Learning"],
    ["Python Programming", "Machine Learning"],
    ["Data Science", "Machine Learning"],
    ["Python Programming", "Data Science", "Machine Learning"]
]

for i, transaction in enumerate(transactions, start=1):
    print(f"Transaction {i}: {transaction}")
```

```python
encoder = TransactionEncoder()
encoded_data = encoder.fit(transactions).transform(transactions)

df = pd.DataFrame(encoded_data, columns=encoder.columns_)

display(df)
```

```python
frequent_itemsets = apriori(
    df,
    min_support=0.4,
    use_colnames=True
)

frequent_itemsets = frequent_itemsets.sort_values(
    by=["support", "itemsets"],
    ascending=[False, True]
).reset_index(drop=True)

display(frequent_itemsets)
```

```python
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.6
)

rules = rules.sort_values(
    by="confidence",
    ascending=False
).reset_index(drop=True)

display(rules[
    ["antecedents", "consequents", "support", "confidence", "lift"]
])
```

```python
if rules.empty:
    print("No association rules satisfy the minimum confidence.")
else:
    result = rules.copy()
    result["antecedents"] = result["antecedents"].apply(
        lambda x: ", ".join(sorted(x))
    )
    result["consequents"] = result["consequents"].apply(
        lambda x: ", ".join(sorted(x))
    )
    result["support"] = (result["support"] * 100).round(2)
    result["confidence"] = (result["confidence"] * 100).round(2)
    result["lift"] = result["lift"].round(2)

    result = result.rename(columns={
        "antecedents": "Books Purchased",
        "consequents": "Books Recommended",
        "support": "Support (%)",
        "confidence": "Confidence (%)",
        "lift": "Lift"
    })

    display(result)
```
